```python
import pandas as pd
import io
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
raw_data_path = '../data/agaricus-lepiota.csv'
```

```python

# 2. Define Column Names in order
column_names = [
    "poisonous", "cap-shape", "cap-surface", "cap-color", "bruises", "odor",
    "gill-attachment", "gill-spacing", "gill-size", "gill-color", "stalk-shape",
    "stalk-root", "stalk-surface-above-ring", "stalk-surface-below-ring",
    "stalk-color-above-ring", "stalk-color-below-ring", "veil-type", "veil-color",
    "ring-number", "ring-type", "spore-print-color", "population", "habitat"
]

# 3. Create DataFrame
df = pd.read_csv(raw_data_path, names=column_names)

print(df.head())
```

```python
for col in df.columns:
    print(f"Value Counts for {col}:")
    print(df[col].value_counts())
    print()
```

```python

# 4. Data Cleaning
df['stalk-root'] = df['stalk-root'].replace('?', 'missing')
df.rename(columns={'poisonous': 'y'}, inplace=True)

# Drop columns with only one unique value (no information gain)
for col in df.columns:
    if df[col].nunique() <= 1:
        df.drop(col, axis=1, inplace=True)

# 5. Encoding
# Label encode the target so it's a single 0/1 column
le = LabelEncoder()
df['y'] = le.fit_transform(df['y'])


# One-Hot Encode only the feature columns
# We identify features by taking all columns except 'y'
features = [col for col in df.columns if col != 'y']
df_processed = pd.get_dummies(df, columns=features, drop_first=True, dtype=int)

# 6. Split and Save
# We split the whole dataframe into two chunks
train_df, test_df = train_test_split(
    df_processed, test_size=0.2, random_state=42, stratify=df_processed['y']
)

# 7. Create Directory and Save
output_dir = '../data'


train_df.to_csv(os.path.join(output_dir, 'mushrooms_preprocessed_train.csv'), index=False)
test_df.to_csv(os.path.join(output_dir, 'mushrooms_preprocessed_test.csv'), index=False)

print(f"Done! Saved complete sets to '{output_dir}/'")
print(f"Train Shape: {train_df.shape} | Test Shape: {test_df.shape}")
```
